# Valores Ausentes (Missing Values)

Nesta seção aprenderemos como identificar e tratar valores ausentes utilizando o Pandas.

```python
import pandas as pd
import numpy as np

dados = {
    "Nome": ["Ana", "Bruno", "Carlos", "Daniela", "Eduardo"],
    "Idade": [23, None, 35, 29, None],
    "Cidade": ["Blumenau", None, "Pomerode", "Indaial", "Gaspar"]
}

df = pd.DataFrame(dados)

df
```

## isna()

A função `isna()` verifica cada posição da tabela e retorna:

- **True** → existe um valor ausente (`NaN`)
- **False** → existe um valor preenchido

Essa função é útil para localizar onde estão os dados faltantes.

```python
df.isna()
```

### Contando valores ausentes

Normalmente não queremos visualizar toda a tabela de `True` e `False`.

Podemos contar quantos valores ausentes existem em cada coluna utilizando:

`isna().sum()`

```python
df.isna().sum()
```

### Visualizando apenas as linhas com idade ausente

Também podemos localizar somente os registros que possuem valores ausentes em uma coluna específica.

```python
df[df["Idade"].isna()]
```

### notna()

A função `notna()` faz exatamente o contrário de `isna()`.

- True → possui valor
- False → está ausente

```python
df[df["Cidade"].notna()]
#df[df["Cidade"].notna() & df["Idade"].notna()]
```

## Exercício

Mostre apenas os registros onde a coluna **Cidade** está vazia.

```python
df[df["Cidade"].isna()]
```

## dropna()

A função `dropna()` remove valores ausentes.

Mas existem várias formas de utilizá-la.

### 1) Sem parâmetros

Remove qualquer linha que possua pelo menos um valor ausente.
Quando usar? Quando todas as colunas são obrigatórias.

```python
df.dropna()
```

### 2) Removendo apenas linhas sem idade

Nem sempre queremos remover todas as linhas.

Podemos indicar quais colunas serão verificadas utilizando `subset`.
Somente linhas onde Idade é NaN serão removidas.

```python
df.dropna(subset=["Idade"])
```

### 3) Verificando mais de uma coluna

Também podemos exigir que determinadas colunas estejam preenchidas.
Remove linhas que possuem valores ausentes em Idade ou Cidade.

```python
df.dropna(subset=["Idade", "Cidade"])
```

### 4) Linhas completamente vazias

Quando importamos um Excel é comum existirem linhas totalmente vazias.


```python
df.dropna(how="all")
```

### 5) Removendo colunas

Também é possível remover colunas que possuem valores ausentes.

axis=0 → trabalha com linhas (padrão)

axis=1 → trabalha com colunas

Na prática, remover colunas inteiras é pouco comum. Geralmente preferimos preencher os valores ausentes.

```python
df.dropna(axis=1)
```

## fillna()

Ao invés de remover registros, podemos preencher os valores ausentes.

```python
df.fillna("Não informado")
```

### Preenchendo apenas uma coluna

```python
df["Cidade"] = df["Cidade"].fillna("Não informado")
```

```python
df
```

### Utilizando a média
Idade é um valor numérico.
Uma estratégia comum é utilizar a média.

```python
media = df["Idade"].mean()

df["Idade"] = df["Idade"].fillna(media)
```

```python
df
```

# Duplicatas

Durante a coleta de dados, é comum que um mesmo registro seja inserido mais de uma vez.

Esses registros repetidos são chamados de **duplicatas**.

Antes de removê-las, devemos identificar se realmente representam um erro ou se fazem parte do processo de negócio.

```python
dados = {
    "Nome": [
        "Ana",
        "Bruno",
        "Carlos",
        "Ana",
        "Eduardo",
        "Carlos"
    ],
    "Cidade": [
        "Blumenau",
        "Pomerode",
        "Gaspar",
        "Blumenau",
        "Indaial",
        "Gaspar"
    ],
    "Idade": [
        25,
        30,
        40,
        25,
        35,
        40
    ]
}

df = pd.DataFrame(dados)

df
```

## duplicated()

A função `duplicated()` verifica quais linhas estão duplicadas.

Ela retorna:

- **False** → primeira ocorrência.
- **True** → registros repetidos.

```python
df.duplicated()
```

## Visualizando apenas as duplicatas

Podemos utilizar o resultado do `duplicated()` como filtro.

```python
df[df.duplicated()]
```

## Contando duplicatas

Para descobrir quantos registros duplicados existem:

```python
df.duplicated().sum()
```

## duplicated(keep="first")
Por padrão (`keep="first"`), a primeira ocorrência é mantida e as demais são marcadas como duplicadas.

```python
df.duplicated(keep="first")
```

Mantém a última ocorrência e marca as anteriores como duplicadas.

```python
df.duplicated(keep="last")
```

## drop_duplicates()

Remove registros duplicados do DataFrame.

Por padrão, mantém a primeira ocorrência.

```python
df.drop_duplicates()
```

### Mantendo a última ocorrência

```python
df.drop_duplicates(keep="last")
```

### Removendo duplicatas considerando apenas uma coluna

Nem sempre precisamos comparar todas as colunas.

Podemos verificar duplicatas utilizando apenas uma ou mais colunas específicas.

```python
df.drop_duplicates(subset=["Nome"])
```

### Quando utilizar?

✔ Quando um cadastro foi importado duas vezes.

✔ Quando um cliente aparece repetido por erro.

✔ Quando registros idênticos foram inseridos acidentalmente.

⚠ Nem toda duplicata deve ser removida.

Exemplo:

Um cliente pode realizar duas compras iguais no mesmo dia.

Nesse caso, os registros são parecidos, mas representam duas vendas diferentes.

# Outliers

Outliers são valores muito diferentes da maioria dos dados.

Eles podem representar:

- erro de digitação;
- erro de coleta;
- ou um valor verdadeiro.

Por isso, **nem todo outlier deve ser removido**.

### Exemplo 1 - Idade

```python
dados = {
    "Idade": [
        18,
        20,
        22,
        25,
        27,
        30,
        150
    ]
}

df = pd.DataFrame(dados)

df
```

Nesse exemplo, a idade **150 anos** provavelmente é um erro de cadastro.

Esse registro deve ser investigado.

### Exemplo 2 — Vendas

```python
dados = {
    "Venda": [
        250,
        300,
        280,
        320,
        350,
        280,
        18000
    ]
}

df = pd.DataFrame(dados)

df
```

O valor **18000** parece um outlier.

Mas será que é um erro?

Talvez não.

Pode representar uma compra corporativa ou um cliente muito importante.

Antes de remover um outlier, sempre verifique o contexto.

## Detectando Outliers com IQR

Uma forma bastante utilizada para detectar outliers é o método **IQR (Intervalo Interquartil)**.

Os passos são:

1. Calcular Q1.
2. Calcular Q3.
3. Calcular o IQR.
4. Definir os limites inferior e superior.
5. Filtrar os valores fora desses limites.

```python
dados = {
    "Idade":[18,20,22,25,27,30,150]
}

df = pd.DataFrame(dados)
```

18 20 22 25 27 30 150
       |
       Q1

18 20 22 25 27 30 150
             |
             Q3

```python
Q1 = df["Idade"].quantile(0.25)

Q3 = df["Idade"].quantile(0.75)

IQR = Q3 - Q1

print(Q1)
print(Q3)
print(IQR) # distancia entre Q1 e Q3
```

```python
limite_inferior = Q1 - 1.5 * IQR

limite_superior = Q3 + 1.5 * IQR

print(limite_inferior)
print(limite_superior)
```

O 1.5 é um fator estatístico definido pelo método do IQR para criar uma margem de segurança ao redor dos dados considerados normais.

```python
outliers = df[
    (df["Idade"] < limite_inferior) |
    (df["Idade"] > limite_superior)
]

outliers
```

## Removendo Outliers

Após identificar os outliers, podemos removê-los.

```python
df = df[
    (df["Idade"] >= limite_inferior) &
    (df["Idade"] <= limite_superior)
]

df

# Ou substituir valor
#df["Idade"] = df["Idade"].replace(150, 50)
```

### Quando remover?

✔ Erros de digitação.

✔ Valores impossíveis.

✔ Dados claramente incorretos.

### Quando manter?

✔ Compras corporativas.

✔ Salário do diretor.

✔ Grandes clientes.

✔ Eventos raros, mas reais.

# Normalização

Normalizar significa padronizar a forma como os dados estão armazenados.

Isso evita problemas durante filtros, agrupamentos e análises.

```python
dados = {
    "Cidade":[
        "Blumenau",
        " blumenau ",
        "BLUMENAU",
        "Blumenau ",
        "Pomerode",
        "POMERODE"
    ]
}

df = pd.DataFrame(dados)

df
```

## strip()
Remove espaços no início e no final do texto.

```python
df["Cidade"] = df["Cidade"].str.strip()

df
```

## lower()
Converte todos os caracteres para minúsculo.

```python
df["Cidade"] = df["Cidade"].str.lower()

df
```

## upper() 
Converte todos os caracteres para maiúsculo.

```python
df["Cidade"] = df["Cidade"].str.upper()

df
```

## title()
Deixa a primeira letra de cada palavra em maiúsculo.

```python
df["Cidade"] = df["Cidade"].str.title()

df
```

## replace()
Substitui um texto por outro.

```python
dados = {
    "Cidade":[
        "BNU",
        "Pomerode",
        "BNU",
        "Gaspar"
    ]
}

df = pd.DataFrame(dados)
```

```python
df["Cidade"] = df["Cidade"].replace(
    "BNU",
    "Blumenau"
)

df
```

## Conversão de tipos
Algumas colunas numéricas podem ser importadas como texto.

Podemos convertê-las utilizando `to_numeric()`.

```python
dados = {
    "Idade":[
        "20",
        "35",
        "abc",
        "40"
    ]
}

df = pd.DataFrame(dados)
```

```python
df["Idade"] = pd.to_numeric(
    df["Idade"],
    errors="coerce" # elimina texto
)

df
```

## Padronização de datas

```python
dados = {
    "Data":[
        "01/02/2025",
        "15/03/2025",
        "20/04/2025"
    ]
}

df = pd.DataFrame(dados)

df
```

```python
df.dtypes
```

```python
df["Data"] = pd.to_datetime(
    df["Data"],
    dayfirst=True # Devido padrão de datas brasileiro 15/01/2026 (Formato no texto: DD/MM/AAAA ou DD-MM-AAAA)
)

# Se as datas já estiverem no padrão internacional ISO (AAAA-MM-DD, como 2026-04-05) não precisa usar dayfirst.

df
```

Após a conversão, a coluna passa a ser do tipo **datetime**, permitindo realizar filtros por datas, extrair mês, ano, dia da semana e outras operações.

```python
# Extrair o ano
print(df["Data"].dt.year)

# Extrair o mês
print(df["Data"].dt.month)

# Extrair o dia
print(df["Data"].dt.day)

# Extrair o nome do dia da semana
print(df["Data"].dt.day_name())
```

```python
# Criar novas colunas no DF
df["Ano"] = df["Data"].dt.year
df["Mes"] = df["Data"].dt.month
df["Dia"] = df["Data"].dt.day
df["Dia_Semana"] = df["Data"].dt.day_name()

print(df)
```

Uma empresa recebeu uma base de clientes exportada de vários sistemas.
